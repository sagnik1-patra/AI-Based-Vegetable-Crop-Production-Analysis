```python
import os
import json
import random
import warnings
from pathlib import Path
from datetime import datetime

import numpy as np
import pandas as pd
import joblib
import yaml

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

warnings.filterwarnings("ignore")

# ============================================================
# 1. PROJECT PATHS
# ============================================================

PROJECT_DIR = Path(
    r"C:\Users\sagni\Downloads\AI-Based Vegetable Crop Production Analysis"
)

DATASET_PATH = PROJECT_DIR / (
    "Production_of_Important_Vegetable_crops_In_Tonnes_2019_20.csv"
)

PROJECT_DIR.mkdir(parents=True, exist_ok=True)

H5_PATH = PROJECT_DIR / "vegetable_model.h5"
PKL_PATH = PROJECT_DIR / "vegetable_model.pkl"
SCALER_PATH = PROJECT_DIR / "vegetable_scaler.pkl"
IMPUTER_PATH = PROJECT_DIR / "vegetable_imputer.pkl"
YAML_PATH = PROJECT_DIR / "config.yaml"
JSON_PATH = PROJECT_DIR / "model_metadata.json"

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

# ============================================================
# 2. LOAD DATASET
# ============================================================

print("=" * 65)
print("AI-BASED VEGETABLE CROP PRODUCTION ANALYSIS")
print("=" * 65)

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found:\n{DATASET_PATH}"
    )

df = pd.read_csv(DATASET_PATH)
df.columns = df.columns.str.strip()

print("\nDataset loaded successfully!")
print("Dataset shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
print("\nFirst five records:")
print(df.head().to_string(index=False))

# ============================================================
# 3. IDENTIFY FEATURES AND TARGET
# ============================================================

TARGET = "Tomato"
LOCATION_COLUMN = "Taluks"

if TARGET not in df.columns:
    raise ValueError(f"Target column '{TARGET}' not found.")

crop_columns = [
    col for col in df.columns
    if col != LOCATION_COLUMN
]

feature_columns = [
    col for col in crop_columns
    if col != TARGET
]

print("\nTarget variable:", TARGET)
print("\nInput features:")
for col in feature_columns:
    print(" -", col)

# ============================================================
# 4. DATA CLEANING
# ============================================================

# Convert crop production columns to numeric.
# Invalid values become NaN and are imputed later.

for col in crop_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Remove records without a known target.
# Never impute target labels for supervised training.

initial_rows = len(df)
df = df.dropna(subset=[TARGET]).copy()

print("\nRemoved rows with missing target:",
      initial_rows - len(df))

if len(df) < 20:
    raise ValueError(
        "Insufficient valid records for model training."
    )

# Production cannot be negative.
for col in crop_columns:
    df.loc[df[col] < 0, col] = np.nan

df = df.dropna(subset=[TARGET]).copy()

print("\nMissing values:")
print(df[feature_columns].isnull().sum())

print("\nTarget statistics:")
print(df[TARGET].describe())

# ============================================================
# 5. TRAIN-TEST SPLIT
# ============================================================

X = df[feature_columns].copy()
y = df[TARGET].astype(float).copy()

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=SEED
)

print("\nTraining samples:", len(X_train))
print("Testing samples:", len(X_test))

# ============================================================
# 6. RANDOM FOREST REGRESSION MODEL
# ============================================================

print("\nTraining Random Forest Regressor...")

rf_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="median")
    ),
    (
        "model",
        RandomForestRegressor(
            n_estimators=200,
            max_depth=10,
            min_samples_leaf=2,
            random_state=SEED,
            n_jobs=-1
        )
    )
])

rf_pipeline.fit(X_train, y_train)

rf_predictions = rf_pipeline.predict(X_test)

# Production estimates should be nonnegative.
rf_predictions = np.maximum(rf_predictions, 0)

# ============================================================
# 7. EVALUATION FUNCTION
# ============================================================

def evaluate_model(y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)

    mse = mean_squared_error(y_true, y_pred)

    rmse = np.sqrt(mse)

    r2 = r2_score(y_true, y_pred)

    return {
        "MAE": float(mae),
        "MSE": float(mse),
        "RMSE": float(rmse),
        "R2": float(r2)
    }


rf_metrics = evaluate_model(
    y_test,
    rf_predictions
)

print("\nRandom Forest Results:")
for metric, value in rf_metrics.items():
    print(f"{metric}: {value:.4f}")

# ============================================================
# 8. PREPROCESSING FOR NEURAL NETWORK
# ============================================================

print("\nPreparing data for neural network...")

# Fit preprocessing ONLY on training data.

imputer = SimpleImputer(strategy="median")

X_train_imputed = imputer.fit_transform(X_train)
X_test_imputed = imputer.transform(X_test)

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(
    X_train_imputed
)

X_test_scaled = scaler.transform(
    X_test_imputed
)

# Log transformation stabilizes highly skewed
# nonnegative production values.

y_train_nn = np.log1p(y_train.to_numpy())
y_test_nn = np.log1p(y_test.to_numpy())

# ============================================================
# 9. BUILD DEEP LEARNING MODEL
# ============================================================

print("\nBuilding Neural Network...")

model = Sequential([
    Input(shape=(X_train_scaled.shape[1],)),

    Dense(64, activation="relu"),
    Dropout(0.20),

    Dense(32, activation="relu"),
    Dropout(0.10),

    Dense(16, activation="relu"),

    Dense(1, activation="linear")
])

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="mse",
    metrics=["mae"]
)

model.summary()

# ============================================================
# 10. TRAIN DEEP LEARNING MODEL
# ============================================================

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=20,
    restore_best_weights=True
)

history = model.fit(
    X_train_scaled,
    y_train_nn,
    validation_split=0.20,
    epochs=200,
    batch_size=16,
    callbacks=[early_stopping],
    verbose=1,
    shuffle=True
)

# ============================================================
# 11. NEURAL NETWORK PREDICTIONS
# ============================================================

nn_log_predictions = model.predict(
    X_test_scaled,
    verbose=0
).flatten()

# Convert log predictions back into tonnes.
# Clip to prevent numerical overflow.

nn_predictions = np.expm1(
    np.clip(nn_log_predictions, -20, 20)
)

nn_predictions = np.maximum(
    nn_predictions, 0
)

nn_metrics = evaluate_model(
    y_test,
    nn_predictions
)

print("\nNeural Network Results:")
for metric, value in nn_metrics.items():
    print(f"{metric}: {value:.4f}")

# ============================================================
# 12. MODEL COMPARISON
# ============================================================

comparison = pd.DataFrame([
    {
        "Model": "Random Forest",
        **rf_metrics
    },
    {
        "Model": "Neural Network",
        **nn_metrics
    }
])

print("\nMODEL PERFORMANCE COMPARISON")
print(comparison.to_string(index=False))

best_model_name = (
    "Random Forest"
    if rf_metrics["RMSE"] <= nn_metrics["RMSE"]
    else "Neural Network"
)

print("\nBest model based on test RMSE:",
      best_model_name)

# ============================================================
# 13. SAVE H5 MODEL
# ============================================================

print("\nSaving neural network model...")

# Legacy HDF5 format as requested.
model.save(str(H5_PATH))

print("Saved:", H5_PATH)

# ============================================================
# 14. SAVE PKL FILES
# ============================================================

print("\nSaving Random Forest model...")

# This pipeline includes the feature imputer.
joblib.dump(
    rf_pipeline,
    PKL_PATH
)

print("Saved:", PKL_PATH)

print("\nSaving scaler...")

joblib.dump(
    scaler,
    SCALER_PATH
)

print("Saved:", SCALER_PATH)

print("\nSaving imputer...")

joblib.dump(
    imputer,
    IMPUTER_PATH
)

print("Saved:", IMPUTER_PATH)

# ============================================================
# 15. SAVE YAML CONFIGURATION
# ============================================================

config = {
    "project": {
        "name": (
            "AI-Based Vegetable Crop "
            "Production Analysis"
        ),
        "dataset": DATASET_PATH.name,
        "year": "2019-20",
        "task": "Regression",
        "target": TARGET,
        "target_unit": "tonnes"
    },

    "data": {
        "total_valid_records": int(len(df)),
        "training_records": int(len(X_train)),
        "testing_records": int(len(X_test)),
        "features": feature_columns,
        "test_size": 0.20,
        "random_state": SEED
    },

    "preprocessing": {
        "missing_value_strategy": "median",
        "scaling": "StandardScaler",
        "target_transformation": "log1p_for_neural_network",
        "negative_production": "treated_as_missing"
    },

    "random_forest": {
        "n_estimators": 200,
        "max_depth": 10,
        "min_samples_leaf": 2,
        "random_state": SEED
    },

    "neural_network": {
        "hidden_layers": [64, 32, 16],
        "activation": "relu",
        "optimizer": "adam",
        "learning_rate": 0.001,
        "epochs": 200,
        "batch_size": 16,
        "early_stopping_patience": 20
    },

    "output_files": {
        "neural_network": H5_PATH.name,
        "random_forest": PKL_PATH.name,
        "scaler": SCALER_PATH.name,
        "imputer": IMPUTER_PATH.name,
        "metadata": JSON_PATH.name
    }
}

with open(YAML_PATH, "w", encoding="utf-8") as f:
    yaml.safe_dump(
        config,
        f,
        sort_keys=False,
        allow_unicode=True
    )

print("\nSaved:", YAML_PATH)

# ============================================================
# 16. SAVE JSON METADATA
# ============================================================

feature_importance = dict(
    zip(
        feature_columns,
        [
            float(value)
            for value in rf_pipeline
            .named_steps["model"]
            .feature_importances_
        ]
    )
)

metadata = {
    "project_name": (
        "AI-Based Vegetable Crop Production Analysis"
    ),

    "created_at": datetime.now().isoformat(),

    "dataset": DATASET_PATH.name,

    "dataset_year": "2019-20",

    "target_variable": TARGET,

    "target_unit": "tonnes",

    "feature_columns": feature_columns,

    "dataset_information": {
        "valid_records": int(len(df)),
        "feature_count": len(feature_columns),
        "train_samples": int(len(X_train)),
        "test_samples": int(len(X_test))
    },

    "model_performance": {
        "Random Forest": rf_metrics,
        "Neural Network": nn_metrics
    },

    "best_model_by_test_rmse": best_model_name,

    "random_forest_feature_importance": (
        feature_importance
    ),

    "neural_network_training": {
        "epochs_completed": len(
            history.history["loss"]
        ),
        "final_training_loss": float(
            history.history["loss"][-1]
        ),
        "final_validation_loss": float(
            history.history["val_loss"][-1]
        )
    },

    "saved_files": {
        "h5": H5_PATH.name,
        "pkl": PKL_PATH.name,
        "scaler": SCALER_PATH.name,
        "imputer": IMPUTER_PATH.name,
        "yaml": YAML_PATH.name
    },

    "limitations": [
        "Single-year production data",
        "No future-year forecasting",
        "Predictions rely on other crop production values",
        "Results are not causal agricultural recommendations"
    ]
}

with open(JSON_PATH, "w", encoding="utf-8") as f:
    json.dump(
        metadata,
        f,
        indent=4,
        ensure_ascii=False
    )

print("\nSaved:", JSON_PATH)

# ============================================================
# 17. VERIFY SAVED FILES
# ============================================================

print("\n" + "=" * 65)
print("VERIFYING GENERATED FILES")
print("=" * 65)

generated_files = [
    H5_PATH,
    PKL_PATH,
    SCALER_PATH,
    IMPUTER_PATH,
    YAML_PATH,
    JSON_PATH
]

for file_path in generated_files:
    if file_path.exists():
        size_kb = file_path.stat().st_size / 1024

        print(
            f"[SUCCESS] {file_path.name} "
            f"({size_kb:.2f} KB)"
        )
    else:
        print(f"[FAILED] {file_path.name}")

# ============================================================
# 18. TEST LOADED RANDOM FOREST MODEL
# ============================================================

print("\nTesting saved Random Forest model...")

loaded_rf = joblib.load(PKL_PATH)

sample_input = X_test.iloc[[0]]

loaded_prediction = loaded_rf.predict(
    sample_input
)[0]

loaded_prediction = max(
    float(loaded_prediction), 0
)

actual_value = float(y_test.iloc[0])

print("\nSample input:")
print(sample_input.to_string(index=False))

print(
    "\nActual Tomato Production:",
    round(actual_value, 2),
    "tonnes"
)

print(
    "Predicted Tomato Production:",
    round(loaded_prediction, 2),
    "tonnes"
)

# ============================================================
# 19. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 65)
print("PROJECT EXECUTION COMPLETED")
print("=" * 65)

print("\nProject directory:")
print(PROJECT_DIR)

print("\nGenerated files:")

for file_path in generated_files:
    print(" -", file_path.name)

print("\nBest model:", best_model_name)

print("\nRandom Forest R2:", round(
    rf_metrics["R2"], 4
))

print("Neural Network R2:", round(
    nn_metrics["R2"], 4
))

print("\nAll requested model artifacts generated!")
```

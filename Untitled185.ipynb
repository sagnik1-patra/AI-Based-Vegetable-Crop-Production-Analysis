```python
import os
import json
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

# ============================================================
# 1. CONFIGURATION
# ============================================================

PROJECT_DIR = Path(
    r"C:\Users\sagni\Downloads\AI-Based Vegetable Crop Production Analysis"
)

DATASET_PATH = PROJECT_DIR / (
    "Production_of_Important_Vegetable_crops_In_Tonnes_2019_20.csv"
)

RF_MODEL_PATH = PROJECT_DIR / "vegetable_model.pkl"
NN_MODEL_PATH = PROJECT_DIR / "vegetable_model.h5"

SCALER_PATH = PROJECT_DIR / "vegetable_scaler.pkl"
IMPUTER_PATH = PROJECT_DIR / "vegetable_imputer.pkl"

METADATA_PATH = PROJECT_DIR / "model_metadata.json"

PROJECT_DIR.mkdir(parents=True, exist_ok=True)

SEED = 42
TARGET = "Tomato"
LOCATION_COLUMN = "Taluks"

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (12, 7)
plt.rcParams["font.size"] = 11

print("=" * 70)
print("AI-BASED VEGETABLE CROP PRODUCTION ANALYSIS")
print("VISUALIZATION AND RESULT GENERATION")
print("=" * 70)

# ============================================================
# 2. CHECK REQUIRED FILES
# ============================================================

required_files = [
    DATASET_PATH,
    RF_MODEL_PATH,
    NN_MODEL_PATH,
    SCALER_PATH,
    IMPUTER_PATH,
    METADATA_PATH
]

for file_path in required_files:
    if not file_path.exists():
        raise FileNotFoundError(
            f"Required file missing:\n{file_path}\n"
            "Run the previous model training code first."
        )

print("\nAll required files found!")

# ============================================================
# 3. LOAD DATASET
# ============================================================

df = pd.read_csv(DATASET_PATH)
df.columns = df.columns.str.strip()

with open(METADATA_PATH, "r", encoding="utf-8") as f:
    metadata = json.load(f)

TARGET = metadata.get("target_variable", TARGET)

feature_columns = metadata["feature_columns"]

crop_columns = feature_columns + [TARGET]

for col in crop_columns:
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

# Match the training script's cleaning procedure.
df = df.dropna(subset=[TARGET]).copy()

for col in crop_columns:
    df.loc[df[col] < 0, col] = np.nan

df = df.dropna(subset=[TARGET]).copy()

print("\nDataset shape:", df.shape)
print("\nTarget:", TARGET)
print("\nFeatures:", feature_columns)

# ============================================================
# 4. LOAD TRAINED MODELS
# ============================================================

print("\nLoading trained models...")

rf_model = joblib.load(RF_MODEL_PATH)

nn_model = tf.keras.models.load_model(
    str(NN_MODEL_PATH),
    compile=False
)

scaler = joblib.load(SCALER_PATH)
imputer = joblib.load(IMPUTER_PATH)

print("Random Forest loaded successfully!")
print("Neural Network loaded successfully!")
print("Scaler loaded successfully!")
print("Imputer loaded successfully!")

# ============================================================
# 5. RECREATE ORIGINAL TEST SPLIT
# ============================================================

X = df[feature_columns].copy()
y = df[TARGET].astype(float).copy()

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=SEED
)

# Keep test-row indices for the CSV files.
test_indices = X_test.index.to_numpy()

# ============================================================
# 6. RANDOM FOREST PREDICTIONS
# ============================================================

rf_predictions = rf_model.predict(X_test)

rf_predictions = np.maximum(
    rf_predictions, 0
)

# ============================================================
# 7. NEURAL NETWORK PREDICTIONS
# ============================================================

X_test_imputed = imputer.transform(X_test)

X_test_scaled = scaler.transform(
    X_test_imputed
)

nn_log_predictions = nn_model.predict(
    X_test_scaled,
    verbose=0
).flatten()

nn_predictions = np.expm1(
    np.clip(nn_log_predictions, -20, 20)
)

nn_predictions = np.maximum(
    nn_predictions, 0
)

actual = y_test.to_numpy()

# ============================================================
# 8. EVALUATION FUNCTION
# ============================================================

def evaluate_model(actual, predicted):

    mae = mean_absolute_error(
        actual, predicted
    )

    mse = mean_squared_error(
        actual, predicted
    )

    rmse = np.sqrt(mse)

    r2 = r2_score(
        actual, predicted
    )

    return {
        "MAE": float(mae),
        "MSE": float(mse),
        "RMSE": float(rmse),
        "R2": float(r2)
    }


rf_metrics = evaluate_model(
    actual,
    rf_predictions
)

nn_metrics = evaluate_model(
    actual,
    nn_predictions
)

print("\nRandom Forest Metrics:")
print(rf_metrics)

print("\nNeural Network Metrics:")
print(nn_metrics)

# ============================================================
# 9. ACCURACY GRAPH (R2 SCORE)
# ============================================================

print("\nGenerating accuracy graph...")

model_names = [
    "Random Forest",
    "Neural Network"
]

r2_values = [
    rf_metrics["R2"],
    nn_metrics["R2"]
]

fig, ax = plt.subplots(figsize=(10, 6))

bars = ax.bar(
    model_names,
    r2_values,
    color=["#3498db", "#2ecc71"],
    width=0.55
)

ax.axhline(
    y=0,
    color="black",
    linewidth=1
)

for bar, value in zip(bars, r2_values):

    offset = 0.02 if value >= 0 else -0.04

    ax.text(
        bar.get_x() + bar.get_width() / 2,
        value + offset,
        f"{value:.4f}",
        ha="center",
        va="bottom" if value >= 0 else "top",
        fontweight="bold",
        fontsize=12
    )

ax.set_title(
    "Vegetable Crop Production - Model R2 Performance",
    fontsize=15,
    fontweight="bold"
)

ax.set_ylabel("R2 Score")
ax.set_xlabel("Machine Learning Models")

ax.grid(axis="y", alpha=0.3)

plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "accuracy_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close(fig)

print("Saved accuracy_graph.png")

# ============================================================
# 10. CROP CORRELATION HEATMAP
# ============================================================

print("\nGenerating correlation heatmap...")

correlation_matrix = df[crop_columns].corr()

fig, ax = plt.subplots(figsize=(13, 10))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    vmin=-1,
    vmax=1,
    linewidths=0.5,
    square=True,
    cbar_kws={"label": "Pearson Correlation"},
    ax=ax
)

ax.set_title(
    "Correlation Heatmap of Vegetable Crop Production",
    fontsize=15,
    fontweight="bold",
    pad=20
)

plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)

plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close(fig)

print("Saved heatmap.png")

# ============================================================
# 11. MODEL COMPARISON GRAPH
# ============================================================

print("\nGenerating model comparison graph...")

comparison_df = pd.DataFrame([
    {
        "Model": "Random Forest",
        **rf_metrics
    },
    {
        "Model": "Neural Network",
        **nn_metrics
    }
])

# Normalize MAE and RMSE separately so their
# relative performance can be visualized together.

metrics_to_plot = ["MAE", "RMSE"]

normalized = comparison_df[
    metrics_to_plot
].copy()

for metric in metrics_to_plot:

    maximum = normalized[metric].max()

    if maximum > 0:
        normalized[metric] /= maximum

x = np.arange(len(metrics_to_plot))
width = 0.35

fig, ax = plt.subplots(figsize=(11, 7))

bars1 = ax.bar(
    x - width / 2,
    normalized.iloc[0].values,
    width,
    label="Random Forest",
    color="#3498db"
)

bars2 = ax.bar(
    x + width / 2,
    normalized.iloc[1].values,
    width,
    label="Neural Network",
    color="#2ecc71"
)

ax.set_xticks(x)
ax.set_xticklabels(metrics_to_plot)

ax.set_ylabel("Normalized Error")
ax.set_xlabel("Evaluation Metrics")

ax.set_title(
    "Machine Learning Model Comparison",
    fontsize=15,
    fontweight="bold"
)

ax.legend()

ax.grid(axis="y", alpha=0.3)

# Show original metric values on bars.
for bars, row in [
    (bars1, comparison_df.iloc[0]),
    (bars2, comparison_df.iloc[1])
]:
    for bar, metric in zip(bars, metrics_to_plot):

        ax.annotate(
            f"{row[metric]:.2f}",
            (
                bar.get_x() + bar.get_width() / 2,
                bar.get_height()
            ),
            xytext=(0, 4),
            textcoords="offset points",
            ha="center",
            fontsize=10
        )

plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "comparison_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close(fig)

print("Saved comparison_graph.png")

# ============================================================
# 12. SELECT BEST MODEL
# ============================================================

# Lower RMSE indicates better test-set performance.

if rf_metrics["RMSE"] <= nn_metrics["RMSE"]:

    best_model_name = "Random Forest"
    best_predictions = rf_predictions
    best_metrics = rf_metrics

else:

    best_model_name = "Neural Network"
    best_predictions = nn_predictions
    best_metrics = nn_metrics

print("\nBest Model:", best_model_name)

# ============================================================
# 13. GENERATE RESULT CSV
# ============================================================

print("\nGenerating result.csv...")

result_df = pd.DataFrame({
    "Taluk": (
        df.loc[test_indices, LOCATION_COLUMN].values
        if LOCATION_COLUMN in df.columns
        else test_indices
    ),

    "Actual_Tomato_Production": actual,

    "Random_Forest_Prediction": rf_predictions,

    "Neural_Network_Prediction": nn_predictions,

    "Random_Forest_Absolute_Error": np.abs(
        actual - rf_predictions
    ),

    "Neural_Network_Absolute_Error": np.abs(
        actual - nn_predictions
    )
})

result_df["Best_Model"] = best_model_name

result_df["Best_Model_Prediction"] = (
    best_predictions
)

result_df["Best_Model_Absolute_Error"] = np.abs(
    actual - best_predictions
)

result_df["Best_Model_Squared_Error"] = (
    actual - best_predictions
) ** 2

result_df.to_csv(
    PROJECT_DIR / "result.csv",
    index=False
)

print("\nResult CSV saved!")
print(result_df.head())

# ============================================================
# 14. RESULT GRAPH
# ============================================================

print("\nGenerating result graph...")

# Sort by actual production for clear visualization.

sorted_results = result_df.sort_values(
    "Actual_Tomato_Production"
).reset_index(drop=True)

fig, ax = plt.subplots(figsize=(14, 7))

ax.plot(
    sorted_results.index,
    sorted_results["Actual_Tomato_Production"],
    label="Actual Production",
    color="black",
    linewidth=2.5
)

ax.plot(
    sorted_results.index,
    sorted_results["Random_Forest_Prediction"],
    label="Random Forest",
    color="#3498db",
    linestyle="--",
    linewidth=2
)

ax.plot(
    sorted_results.index,
    sorted_results["Neural_Network_Prediction"],
    label="Neural Network",
    color="#2ecc71",
    linestyle=":",
    linewidth=2
)

ax.set_title(
    "Actual vs Estimated Tomato Production",
    fontsize=15,
    fontweight="bold"
)

ax.set_xlabel(
    "Test Samples Sorted by Actual Production"
)

ax.set_ylabel("Tomato Production (Tonnes)")

ax.legend()
ax.grid(alpha=0.3)

plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "result_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close(fig)

print("Saved result_graph.png")

# ============================================================
# 15. GENERATE PREDICTION CSV
# ============================================================

print("\nGenerating prediction.csv...")

prediction_df = X_test.copy()

prediction_df.insert(
    0,
    "Taluk",
    (
        df.loc[test_indices, LOCATION_COLUMN].values
        if LOCATION_COLUMN in df.columns
        else test_indices
    )
)

prediction_df["Actual_Tomato_Production"] = actual

prediction_df["Predicted_Tomato_Production"] = (
    best_predictions
)

prediction_df["Prediction_Error"] = (
    actual - best_predictions
)

prediction_df["Absolute_Error"] = np.abs(
    prediction_df["Prediction_Error"]
)

prediction_df["Model_Used"] = best_model_name

prediction_df.to_csv(
    PROJECT_DIR / "prediction.csv",
    index=False
)

print("\nPrediction CSV saved!")
print(prediction_df.head())

# ============================================================
# 16. PREDICTION GRAPH
# ============================================================

print("\nGenerating prediction graph...")

fig, ax = plt.subplots(figsize=(11, 8))

ax.scatter(
    actual,
    best_predictions,
    color="#3498db",
    alpha=0.75,
    s=65,
    edgecolors="black",
    linewidths=0.4
)

minimum = min(
    np.min(actual),
    np.min(best_predictions)
)

maximum = max(
    np.max(actual),
    np.max(best_predictions)
)

ax.plot(
    [minimum, maximum],
    [minimum, maximum],
    color="red",
    linestyle="--",
    linewidth=2,
    label="Perfect Prediction"
)

ax.set_xlabel(
    "Actual Tomato Production (Tonnes)"
)

ax.set_ylabel(
    "Predicted Tomato Production (Tonnes)"
)

ax.set_title(
    f"Prediction Performance - {best_model_name}",
    fontsize=15,
    fontweight="bold"
)

ax.text(
    0.05,
    0.95,
    f"R2 = {best_metrics['R2']:.4f}\n"
    f"RMSE = {best_metrics['RMSE']:.2f}\n"
    f"MAE = {best_metrics['MAE']:.2f}",
    transform=ax.transAxes,
    fontsize=11,
    verticalalignment="top",
    bbox={
        "boxstyle": "round",
        "facecolor": "white",
        "alpha": 0.85
    }
)

ax.legend()
ax.grid(alpha=0.3)

plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "prediction_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close(fig)

print("Saved prediction_graph.png")

# ============================================================
# 17. ADDITIONAL: FEATURE IMPORTANCE GRAPH
# ============================================================

print("\nGenerating feature importance graph...")

rf_estimator = rf_model.named_steps["model"]

feature_importance = pd.DataFrame({
    "Feature": feature_columns,
    "Importance": rf_estimator.feature_importances_
})

feature_importance = feature_importance.sort_values(
    "Importance",
    ascending=True
)

fig, ax = plt.subplots(figsize=(11, 7))

ax.barh(
    feature_importance["Feature"],
    feature_importance["Importance"],
    color="#9b59b6"
)

ax.set_title(
    "Random Forest Feature Importance",
    fontsize=15,
    fontweight="bold"
)

ax.set_xlabel("Feature Importance")

plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "feature_importance.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close(fig)

print("Saved feature_importance.png")

# ============================================================
# 18. SAVE MODEL COMPARISON CSV
# ============================================================

comparison_df.to_csv(
    PROJECT_DIR / "model_comparison.csv",
    index=False
)

print("\nSaved model_comparison.csv")

# ============================================================
# 19. FINAL PERFORMANCE REPORT
# ============================================================

print("\n" + "=" * 70)
print("FINAL MODEL PERFORMANCE")
print("=" * 70)

print(comparison_df.to_string(index=False))

print("\nBest Model:", best_model_name)

print("\nBest Model Metrics:")

for metric, value in best_metrics.items():
    print(f"{metric}: {value:.4f}")

# ============================================================
# 20. VERIFY ALL OUTPUT FILES
# ============================================================

output_files = [
    "accuracy_graph.png",
    "heatmap.png",
    "comparison_graph.png",
    "result.csv",
    "result_graph.png",
    "prediction.csv",
    "prediction_graph.png",
    "feature_importance.png",
    "model_comparison.csv"
]

print("\n" + "=" * 70)
print("GENERATED OUTPUT FILES")
print("=" * 70)

for filename in output_files:

    file_path = PROJECT_DIR / filename

    if file_path.exists():

        file_size = (
            file_path.stat().st_size / 1024
        )

        print(
            f"[SUCCESS] {filename} "
            f"({file_size:.2f} KB)"
        )

    else:
        print(f"[FAILED] {filename}")

print("\n" + "=" * 70)
print("ALL VISUALIZATIONS AND CSV FILES GENERATED")
print("=" * 70)

print("\nOutput Directory:")
print(PROJECT_DIR)

print("\nProject completed successfully!")
```

```python

# ============================================================
# AI-BASED VEGETABLE CROP PRODUCTION ANALYSIS
# ARTIFICIAL IMMUNE SYSTEM (AIS) - CLONALG
#
# Outputs use ais_ prefix
# ============================================================

import os
import json
import random
import warnings
from pathlib import Path
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import yaml
import tensorflow as tf

from sklearn.model_selection import (
    train_test_split,
    KFold,
    cross_val_score
)
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

warnings.filterwarnings("ignore")

# ============================================================
# 1. CONFIGURATION
# ============================================================

PROJECT_DIR = Path(
    r"C:\Users\sagni\Downloads\AI-Based Vegetable Crop Production Analysis"
)

DATASET_PATH = PROJECT_DIR / (
    "Production_of_Important_Vegetable_crops_In_Tonnes_2019_20.csv"
)

PROJECT_DIR.mkdir(parents=True, exist_ok=True)

SEED = 42
TARGET = "Tomato"
LOCATION_COLUMN = "Taluks"

POPULATION_SIZE = 12
GENERATIONS = 12
CLONES_PER_PARENT = 4
ELITE_COUNT = 4
MUTATION_RATE = 0.25

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

sns.set_theme(style="whitegrid")

plt.rcParams["figure.figsize"] = (12, 7)
plt.rcParams["font.size"] = 11

def output_path(filename):
    return PROJECT_DIR / ("ais_" + filename)

print("=" * 70)
print("ARTIFICIAL IMMUNE SYSTEM")
print("VEGETABLE CROP PRODUCTION ANALYSIS")
print("=" * 70)

# ============================================================
# 2. LOAD AND CLEAN DATASET
# ============================================================

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found: {DATASET_PATH}"
    )

df = pd.read_csv(DATASET_PATH)
df.columns = df.columns.str.strip()

if TARGET not in df.columns:
    raise ValueError(
        f"Target column '{TARGET}' not found.\n"
        f"Available columns: {df.columns.tolist()}"
    )

crop_columns = [
    c for c in df.columns
    if c != LOCATION_COLUMN
]

feature_columns = [
    c for c in crop_columns
    if c != TARGET
]

for col in crop_columns:
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

# Match baseline cleaning.
df = df.dropna(subset=[TARGET]).copy()

for col in crop_columns:
    df.loc[df[col] < 0, col] = np.nan

df = df.dropna(subset=[TARGET]).copy()

X = df[feature_columns].copy()
y = df[TARGET].astype(float).copy()

print("\nDataset shape:", df.shape)
print("Target:", TARGET)
print("Features:", feature_columns)

# ============================================================
# 3. TRAIN-TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=SEED
)

print("\nTraining records:", len(X_train))
print("Testing records:", len(X_test))

# ============================================================
# 4. AIS ANTIBODY REPRESENTATION
# ============================================================

# Each antibody contains:
# - Random Forest hyperparameters
# - Binary feature selection mask

N_FEATURES = len(feature_columns)

def create_antibody():

    mask = np.random.randint(
        0, 2, size=N_FEATURES
    ).tolist()

    if sum(mask) == 0:
        mask[np.random.randint(N_FEATURES)] = 1

    return {
        "n_estimators": random.randint(50, 250),
        "max_depth": random.randint(3, 20),
        "min_samples_split": random.randint(2, 12),
        "min_samples_leaf": random.randint(1, 6),
        "max_features": random.uniform(0.4, 1.0),
        "feature_mask": mask
    }


def selected_features(antibody):

    return [
        feature_columns[i]
        for i, flag in enumerate(
            antibody["feature_mask"]
        )
        if flag == 1
    ]

# ============================================================
# 5. ANTIBODY FITNESS FUNCTION
# ============================================================

fitness_cache = {}

cv = KFold(
    n_splits=3,
    shuffle=True,
    random_state=SEED
)

def antibody_key(antibody):

    return (
        antibody["n_estimators"],
        antibody["max_depth"],
        antibody["min_samples_split"],
        antibody["min_samples_leaf"],
        round(antibody["max_features"], 4),
        tuple(antibody["feature_mask"])
    )


def calculate_fitness(antibody):

    key = antibody_key(antibody)

    if key in fitness_cache:
        return fitness_cache[key]

    features = selected_features(antibody)

    if not features:
        return -1e12

    model = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "rf",
            RandomForestRegressor(
                n_estimators=antibody["n_estimators"],
                max_depth=antibody["max_depth"],
                min_samples_split=antibody[
                    "min_samples_split"
                ],
                min_samples_leaf=antibody[
                    "min_samples_leaf"
                ],
                max_features=antibody[
                    "max_features"
                ],
                random_state=SEED,
                n_jobs=1
            )
        )
    ])

    try:
        scores = cross_val_score(
            model,
            X_train[features],
            y_train,
            cv=cv,
            scoring="neg_root_mean_squared_error",
            n_jobs=-1,
            error_score="raise"
        )

        fitness = float(np.mean(scores))

    except Exception:
        fitness = -1e12

    fitness_cache[key] = fitness

    return fitness

# ============================================================
# 6. CLONAL MUTATION
# ============================================================

def mutate_antibody(antibody, mutation_rate):

    child = {
        k: (
            v.copy() if isinstance(v, list) else v
        )
        for k, v in antibody.items()
    }

    if random.random() < mutation_rate:
        child["n_estimators"] = int(
            np.clip(
                child["n_estimators"] +
                random.randint(-40, 40),
                30,
                300
            )
        )

    if random.random() < mutation_rate:
        child["max_depth"] = int(
            np.clip(
                child["max_depth"] +
                random.randint(-4, 4),
                2,
                25
            )
        )

    if random.random() < mutation_rate:
        child["min_samples_split"] = int(
            np.clip(
                child["min_samples_split"] +
                random.randint(-3, 3),
                2,
                15
            )
        )

    if random.random() < mutation_rate:
        child["min_samples_leaf"] = int(
            np.clip(
                child["min_samples_leaf"] +
                random.randint(-2, 2),
                1,
                8
            )
        )

    if random.random() < mutation_rate:
        child["max_features"] = float(
            np.clip(
                child["max_features"] +
                random.uniform(-0.2, 0.2),
                0.3,
                1.0
            )
        )

    for i in range(N_FEATURES):

        if random.random() < mutation_rate / 2:
            child["feature_mask"][i] ^= 1

    if sum(child["feature_mask"]) == 0:
        child["feature_mask"][
            random.randrange(N_FEATURES)
        ] = 1

    return child

# ============================================================
# 7. AIS OPTIMIZATION
# ============================================================

print("\nStarting AIS optimization...")

population = [
    create_antibody()
    for _ in range(POPULATION_SIZE)
]

best_antibody = None
best_fitness = -np.inf

convergence_history = []

for generation in range(1, GENERATIONS + 1):

    scored_population = [
        (calculate_fitness(a), a)
        for a in population
    ]

    scored_population.sort(
        key=lambda x: x[0],
        reverse=True
    )

    current_fitness, current_best = (
        scored_population[0]
    )

    if current_fitness > best_fitness:

        best_fitness = current_fitness

        best_antibody = {
            k: (
                v.copy() if isinstance(v, list)
                else v
            )
            for k, v in current_best.items()
        }

    convergence_history.append(
        -best_fitness
    )

    print(
        f"Generation {generation:02d}/{GENERATIONS} "
        f"| Best CV RMSE: {-best_fitness:.4f}"
    )

    # Preserve elite antibodies.
    elites = [
        item[1]
        for item in scored_population[:ELITE_COUNT]
    ]

    clones = []

    for rank, parent in enumerate(elites):

        # Higher-ranked antibodies receive
        # lower mutation intensity.
        mutation_strength = (
            MUTATION_RATE * (1 + rank * 0.3)
        )

        for _ in range(CLONES_PER_PARENT):

            clone = mutate_antibody(
                parent,
                mutation_strength
            )

            clones.append(clone)

    candidates = (
        [item[1] for item in scored_population]
        + clones
    )

    candidates.sort(
        key=calculate_fitness,
        reverse=True
    )

    # Keep strongest antibodies and introduce
    # new random antibodies for diversity.
    survivors = candidates[
        :POPULATION_SIZE - 2
    ]

    population = survivors + [
        create_antibody(),
        create_antibody()
    ]

print("\nAIS optimization completed!")

print("\nBest antibody:")
print(best_antibody)

print("\nBest CV RMSE:", -best_fitness)

# ============================================================
# 8. TRAIN AIS-OPTIMIZED RANDOM FOREST
# ============================================================

ais_features = selected_features(
    best_antibody
)

print("\nSelected features:")
print(ais_features)

ais_rf_model = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="median")
    ),
    (
        "rf",
        RandomForestRegressor(
            n_estimators=best_antibody[
                "n_estimators"
            ],
            max_depth=best_antibody[
                "max_depth"
            ],
            min_samples_split=best_antibody[
                "min_samples_split"
            ],
            min_samples_leaf=best_antibody[
                "min_samples_leaf"
            ],
            max_features=best_antibody[
                "max_features"
            ],
            random_state=SEED,
            n_jobs=-1
        )
    )
])

ais_rf_model.fit(
    X_train[ais_features],
    y_train
)

rf_predictions = np.maximum(
    ais_rf_model.predict(
        X_test[ais_features]
    ),
    0
)

# ============================================================
# 9. AIS-GUIDED NEURAL NETWORK
# ============================================================

# The neural network uses the features selected
# by AIS. Its architecture is not optimized by AIS.

imputer = SimpleImputer(strategy="median")

X_train_imputed = imputer.fit_transform(
    X_train[ais_features]
)

X_test_imputed = imputer.transform(
    X_test[ais_features]
)

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(
    X_train_imputed
)

X_test_scaled = scaler.transform(
    X_test_imputed
)

y_train_log = np.log1p(
    y_train.to_numpy()
)

ais_nn_model = Sequential([
    Input(shape=(len(ais_features),)),
    Dense(64, activation="relu"),
    Dropout(0.20),
    Dense(32, activation="relu"),
    Dropout(0.10),
    Dense(16, activation="relu"),
    Dense(1)
])

ais_nn_model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="mse",
    metrics=["mae"]
)

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=20,
    restore_best_weights=True
)

history = ais_nn_model.fit(
    X_train_scaled,
    y_train_log,
    validation_split=0.20,
    epochs=200,
    batch_size=16,
    callbacks=[early_stopping],
    verbose=0
)

nn_log_predictions = ais_nn_model.predict(
    X_test_scaled,
    verbose=0
).flatten()

nn_predictions = np.maximum(
    np.expm1(
        np.clip(nn_log_predictions, -20, 20)
    ),
    0
)

# ============================================================
# 10. EVALUATION
# ============================================================

def evaluate(actual, predicted):

    return {
        "MAE": float(
            mean_absolute_error(actual, predicted)
        ),
        "MSE": float(
            mean_squared_error(actual, predicted)
        ),
        "RMSE": float(
            np.sqrt(
                mean_squared_error(
                    actual,
                    predicted
                )
            )
        ),
        "R2": float(
            r2_score(actual, predicted)
        )
    }


actual = y_test.to_numpy()

rf_metrics = evaluate(
    actual,
    rf_predictions
)

nn_metrics = evaluate(
    actual,
    nn_predictions
)

comparison_df = pd.DataFrame([
    {
        "Model": "AIS-Random Forest",
        **rf_metrics
    },
    {
        "Model": "AIS-Guided Neural Network",
        **nn_metrics
    }
])

comparison_df.to_csv(
    output_path("model_comparison.csv"),
    index=False
)

print("\nAIS MODEL COMPARISON")
print(comparison_df.to_string(index=False))

if rf_metrics["RMSE"] <= nn_metrics["RMSE"]:

    best_model_name = "AIS-Random Forest"
    best_predictions = rf_predictions
    best_metrics = rf_metrics

else:

    best_model_name = "AIS-Guided Neural Network"
    best_predictions = nn_predictions
    best_metrics = nn_metrics

print("\nBest model:", best_model_name)

# ============================================================
# 11. SAVE MODELS
# ============================================================

joblib.dump(
    {
        "pipeline": ais_rf_model,
        "selected_features": ais_features,
        "target": TARGET
    },
    output_path("vegetable_model.pkl")
)

joblib.dump(
    scaler,
    output_path("vegetable_scaler.pkl")
)

joblib.dump(
    imputer,
    output_path("vegetable_imputer.pkl")
)

ais_nn_model.save(
    str(output_path("vegetable_model.h5"))
)

print("\nAIS model artifacts saved!")

# ============================================================
# 12. SAVE YAML CONFIGURATION
# ============================================================

config = {
    "project_name": (
        "AI-Based Vegetable Crop Production Analysis"
    ),
    "optimization": "Artificial Immune System",
    "algorithm": "Clonal Selection Algorithm",
    "target": TARGET,
    "target_unit": "tonnes",
    "dataset": DATASET_PATH.name,
    "selected_features": ais_features,
    "ais_parameters": {
        "population_size": POPULATION_SIZE,
        "generations": GENERATIONS,
        "clones_per_parent": CLONES_PER_PARENT,
        "elite_count": ELITE_COUNT,
        "mutation_rate": MUTATION_RATE
    },
    "best_hyperparameters": best_antibody,
    "best_cv_rmse": float(-best_fitness),
    "random_state": SEED
}

with open(
    output_path("config.yaml"),
    "w",
    encoding="utf-8"
) as f:

    yaml.safe_dump(
        config,
        f,
        sort_keys=False
    )

# ============================================================
# 13. SAVE JSON METADATA
# ============================================================

metadata = {
    "project": (
        "AI-Based Vegetable Crop Production Analysis"
    ),
    "optimization_method": "AIS-CLONALG",
    "created_at": datetime.now().isoformat(),
    "target": TARGET,
    "selected_features": ais_features,
    "best_antibody": best_antibody,
    "cross_validation_rmse": float(-best_fitness),
    "model_performance": {
        "AIS-Random Forest": rf_metrics,
        "AIS-Guided Neural Network": nn_metrics
    },
    "best_model": best_model_name,
    "convergence_rmse": convergence_history,
    "training_samples": int(len(X_train)),
    "testing_samples": int(len(X_test)),
    "limitations": [
        "Single-year production data",
        "No temporal forecasting",
        "AIS optimizes Random Forest and feature selection",
        "Neural network uses AIS-selected features"
    ]
}

with open(
    output_path("model_metadata.json"),
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        metadata,
        f,
        indent=4
    )

# ============================================================
# 14. AIS ACCURACY GRAPH (R2)
# ============================================================

fig, ax = plt.subplots(figsize=(10, 6))

names = [
    "AIS-Random Forest",
    "AIS-Guided NN"
]

scores = [
    rf_metrics["R2"],
    nn_metrics["R2"]
]

bars = ax.bar(
    names,
    scores,
    color=["#2980b9", "#27ae60"]
)

ax.axhline(0, color="black", linewidth=1)

for bar, score in zip(bars, scores):

    ax.annotate(
        f"{score:.4f}",
        (
            bar.get_x() + bar.get_width() / 2,
            bar.get_height()
        ),
        xytext=(
            0,
            5 if score >= 0 else -15
        ),
        textcoords="offset points",
        ha="center",
        fontsize=11,
        fontweight="bold"
    )

ax.set_title(
    "AIS Model Performance (R2 Score)",
    fontweight="bold"
)

ax.set_ylabel("R2 Score")
ax.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.savefig(
    output_path("accuracy_graph.png"),
    dpi=300,
    bbox_inches="tight"
)
plt.show()
plt.close(fig)

# ============================================================
# 15. AIS HEATMAP
# ============================================================

fig, ax = plt.subplots(figsize=(13, 10))

sns.heatmap(
    df[crop_columns].corr(),
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    vmin=-1,
    vmax=1,
    linewidths=0.5,
    ax=ax
)

ax.set_title(
    "AIS Study - Vegetable Production Correlation",
    fontweight="bold"
)

plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)

plt.tight_layout()
plt.savefig(
    output_path("heatmap.png"),
    dpi=300,
    bbox_inches="tight"
)
plt.show()
plt.close(fig)

# ============================================================
# 16. AIS MODEL COMPARISON GRAPH
# ============================================================

metrics = ["MAE", "RMSE"]

fig, ax = plt.subplots(figsize=(11, 7))

x = np.arange(len(metrics))
width = 0.35

rf_values = [
    rf_metrics[m] for m in metrics
]

nn_values = [
    nn_metrics[m] for m in metrics
]

bars1 = ax.bar(
    x - width / 2,
    rf_values,
    width,
    label="AIS-Random Forest",
    color="#3498db"
)

bars2 = ax.bar(
    x + width / 2,
    nn_values,
    width,
    label="AIS-Guided NN",
    color="#2ecc71"
)

ax.set_xticks(x)
ax.set_xticklabels(metrics)

ax.set_ylabel("Prediction Error (Tonnes)")
ax.set_title(
    "AIS Model Comparison - MAE and RMSE",
    fontweight="bold"
)

ax.legend()
ax.grid(axis="y", alpha=0.3)

for bars in [bars1, bars2]:
    for bar in bars:
        ax.annotate(
            f"{bar.get_height():.2f}",
            (
                bar.get_x() + bar.get_width() / 2,
                bar.get_height()
            ),
            xytext=(0, 4),
            textcoords="offset points",
            ha="center",
            fontsize=9
        )

plt.tight_layout()
plt.savefig(
    output_path("comparison_graph.png"),
    dpi=300,
    bbox_inches="tight"
)
plt.show()
plt.close(fig)

# ============================================================
# 17. AIS RESULT CSV
# ============================================================

test_indices = X_test.index

if LOCATION_COLUMN in df.columns:

    taluks = df.loc[
        test_indices,
        LOCATION_COLUMN
    ].values

else:
    taluks = test_indices.to_numpy()

result_df = pd.DataFrame({
    "Taluk": taluks,
    "Actual_Tomato_Production": actual,
    "AIS_RF_Prediction": rf_predictions,
    "AIS_NN_Prediction": nn_predictions,
    "AIS_RF_Absolute_Error": np.abs(
        actual - rf_predictions
    ),
    "AIS_NN_Absolute_Error": np.abs(
        actual - nn_predictions
    ),
    "Best_Model": best_model_name,
    "Best_Model_Prediction": best_predictions,
    "Best_Model_Absolute_Error": np.abs(
        actual - best_predictions
    )
})

result_df.to_csv(
    output_path("result.csv"),
    index=False
)

print("\nAIS Result CSV:")
print(result_df.head())

# ============================================================
# 18. AIS RESULT GRAPH
# ============================================================

sorted_results = result_df.sort_values(
    "Actual_Tomato_Production"
).reset_index(drop=True)

fig, ax = plt.subplots(figsize=(14, 7))

ax.plot(
    sorted_results.index,
    sorted_results["Actual_Tomato_Production"],
    label="Actual",
    color="black",
    linewidth=2.5
)

ax.plot(
    sorted_results.index,
    sorted_results["AIS_RF_Prediction"],
    label="AIS-Random Forest",
    linestyle="--",
    color="#3498db"
)

ax.plot(
    sorted_results.index,
    sorted_results["AIS_NN_Prediction"],
    label="AIS-Guided NN",
    linestyle=":",
    color="#2ecc71"
)

ax.set_title(
    "AIS Actual vs Estimated Tomato Production",
    fontweight="bold"
)

ax.set_xlabel(
    "Test Samples Sorted by Actual Production"
)
ax.set_ylabel("Production (Tonnes)")
ax.legend()
ax.grid(alpha=0.3)

plt.tight_layout()
plt.savefig(
    output_path("result_graph.png"),
    dpi=300,
    bbox_inches="tight"
)
plt.show()
plt.close(fig)

# ============================================================
# 19. AIS PREDICTION CSV
# ============================================================

prediction_df = X_test.copy()

prediction_df.insert(
    0,
    "Taluk",
    taluks
)

prediction_df["Actual_Tomato_Production"] = actual

prediction_df["Predicted_Tomato_Production"] = (
    best_predictions
)

prediction_df["Prediction_Error"] = (
    actual - best_predictions
)

prediction_df["Absolute_Error"] = np.abs(
    prediction_df["Prediction_Error"]
)

prediction_df["Model_Used"] = best_model_name

prediction_df.to_csv(
    output_path("prediction.csv"),
    index=False
)

print("\nAIS Prediction CSV:")
print(prediction_df.head())

# ============================================================
# 20. AIS PREDICTION GRAPH
# ============================================================

fig, ax = plt.subplots(figsize=(11, 8))

ax.scatter(
    actual,
    best_predictions,
    color="#2980b9",
    s=65,
    alpha=0.75,
    edgecolors="black",
    linewidths=0.4
)

minimum = min(
    actual.min(),
    best_predictions.min()
)

maximum = max(
    actual.max(),
    best_predictions.max()
)

ax.plot(
    [minimum, maximum],
    [minimum, maximum],
    "r--",
    linewidth=2,
    label="Perfect Prediction"
)

ax.set_title(
    f"AIS Prediction Performance - {best_model_name}",
    fontweight="bold"
)

ax.set_xlabel("Actual Production (Tonnes)")
ax.set_ylabel("Predicted Production (Tonnes)")

ax.text(
    0.05,
    0.95,
    f"R2: {best_metrics['R2']:.4f}\n"
    f"RMSE: {best_metrics['RMSE']:.2f}\n"
    f"MAE: {best_metrics['MAE']:.2f}",
    transform=ax.transAxes,
    va="top",
    bbox={
        "boxstyle": "round",
        "facecolor": "white",
        "alpha": 0.85
    }
)

ax.legend()
ax.grid(alpha=0.3)

plt.tight_layout()
plt.savefig(
    output_path("prediction_graph.png"),
    dpi=300,
    bbox_inches="tight"
)
plt.show()
plt.close(fig)

# ============================================================
# 21. AIS FEATURE IMPORTANCE
# ============================================================

importance = ais_rf_model.named_steps[
    "rf"
].feature_importances_

importance_df = pd.DataFrame({
    "Feature": ais_features,
    "Importance": importance
}).sort_values("Importance")

fig, ax = plt.subplots(figsize=(11, 7))

ax.barh(
    importance_df["Feature"],
    importance_df["Importance"],
    color="#8e44ad"
)

ax.set_title(
    "AIS-Selected Feature Importance",
    fontweight="bold"
)

ax.set_xlabel("Random Forest Feature Importance")

plt.tight_layout()
plt.savefig(
    output_path("feature_importance.png"),
    dpi=300,
    bbox_inches="tight"
)
plt.show()
plt.close(fig)

# ============================================================
# 22. AIS CONVERGENCE GRAPH
# ============================================================

fig, ax = plt.subplots(figsize=(11, 7))

ax.plot(
    range(1, GENERATIONS + 1),
    convergence_history,
    marker="o",
    color="#e67e22",
    linewidth=2
)

ax.set_title(
    "AIS Optimization Convergence",
    fontweight="bold"
)

ax.set_xlabel("Generation")
ax.set_ylabel("Best Cross-Validation RMSE")

ax.grid(alpha=0.3)

plt.tight_layout()
plt.savefig(
    output_path("convergence_graph.png"),
    dpi=300,
    bbox_inches="tight"
)
plt.show()
plt.close(fig)

# ============================================================
# 23. VERIFY OUTPUTS
# ============================================================

output_files = [
    "vegetable_model.h5",
    "vegetable_model.pkl",
    "vegetable_scaler.pkl",
    "vegetable_imputer.pkl",
    "config.yaml",
    "model_metadata.json",
    "accuracy_graph.png",
    "heatmap.png",
    "comparison_graph.png",
    "result.csv",
    "result_graph.png",
    "prediction.csv",
    "prediction_graph.png",
    "feature_importance.png",
    "convergence_graph.png",
    "model_comparison.csv"
]

print("\n" + "=" * 70)
print("AIS GENERATED OUTPUT FILES")
print("=" * 70)

for filename in output_files:

    path = output_path(filename)

    if path.exists():

        size_kb = path.stat().st_size / 1024

        print(
            f"[SUCCESS] {path.name} "
            f"({size_kb:.2f} KB)"
        )

    else:
        print(f"[FAILED] {path.name}")

print("\n" + "=" * 70)
print("AIS OPTIMIZATION COMPLETED SUCCESSFULLY")
print("=" * 70)

print("\nBest AIS Model:", best_model_name)
print("Best CV RMSE:", round(-best_fitness, 4))
print("Test R2:", round(best_metrics["R2"], 4))
print("Test RMSE:", round(best_metrics["RMSE"], 4))

print("\nAll AIS files saved in:")
print(PROJECT_DIR)
```

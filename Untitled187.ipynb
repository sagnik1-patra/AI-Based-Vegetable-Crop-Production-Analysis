```python
import os
import json
import random
import warnings
from pathlib import Path
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import yaml
import tensorflow as tf

from sklearn.model_selection import (
    train_test_split,
    KFold,
    cross_val_score
)
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

warnings.filterwarnings("ignore")

# ================================================================
# 1. CONFIGURATION
# ================================================================

PROJECT_DIR = Path(
    r"C:\Users\sagni\Downloads\AI-Based Vegetable Crop Production Analysis"
)

DATASET_PATH = PROJECT_DIR / (
    "Production_of_Important_Vegetable_crops_In_Tonnes_2019_20.csv"
)

PROJECT_DIR.mkdir(parents=True, exist_ok=True)

SEED = 42
TARGET = "Tomato"
LOCATION_COLUMN = "Taluks"

# PSO parameters
NUM_PARTICLES = 12
MAX_ITERATIONS = 15
INERTIA = 0.7
COGNITIVE = 1.5
SOCIAL = 1.5

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

rng = np.random.default_rng(SEED)

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (12, 7)
plt.rcParams["font.size"] = 11

def out(name):
    return PROJECT_DIR / ("pso_" + name)

def save_plot(filename):
    plt.tight_layout()
    plt.savefig(
        out(filename),
        dpi=300,
        bbox_inches="tight"
    )
    plt.show()
    plt.close()

print("=" * 70)
print("AI-BASED VEGETABLE CROP PRODUCTION ANALYSIS")
print("PARTICLE SWARM OPTIMIZATION")
print("=" * 70)

# ================================================================
# 2. LOAD AND PREPROCESS DATA
# ================================================================

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found:\n{DATASET_PATH}"
    )

df = pd.read_csv(DATASET_PATH)
df.columns = df.columns.str.strip()

if TARGET not in df.columns:
    raise ValueError(
        f"Target '{TARGET}' not found. "
        f"Available columns: {df.columns.tolist()}"
    )

crop_columns = [
    c for c in df.columns
    if c != LOCATION_COLUMN
]

feature_columns = [
    c for c in crop_columns
    if c != TARGET
]

for col in crop_columns:
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

# Same cleaning as baseline and AIS.
df = df.dropna(subset=[TARGET]).copy()

for col in crop_columns:
    df.loc[df[col] < 0, col] = np.nan

df = df.dropna(subset=[TARGET]).copy()

X = df[feature_columns].copy()
y = df[TARGET].astype(float).copy()

print("\nDataset shape:", df.shape)
print("Target:", TARGET)
print("Features:", feature_columns)
print("\nMissing values:")
print(df[crop_columns].isna().sum())

# ================================================================
# 3. CREATE TRAIN, VALIDATION AND TEST SPLITS
# ================================================================

# Keep the same 80:20 outer split as the baseline/AIS.
X_train_full, X_test, y_train_full, y_test = (
    train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=SEED
    )
)

# Separate validation data for selecting the final model.
X_train, X_val, y_train, y_val = train_test_split(
    X_train_full,
    y_train_full,
    test_size=0.20,
    random_state=SEED
)

print("\nTraining records:", len(X_train))
print("Validation records:", len(X_val))
print("Testing records:", len(X_test))

# ================================================================
# 4. PARTICLE REPRESENTATION
# ================================================================

# Particle dimensions:
# 0: n_estimators
# 1: max_depth
# 2: min_samples_split
# 3: min_samples_leaf
# 4: max_features
# 5 onward: feature selection values

N_FEATURES = len(feature_columns)
DIMENSIONS = 5 + N_FEATURES

LOWER_BOUNDS = np.array(
    [50, 3, 2, 1, 0.4] + [0.0] * N_FEATURES,
    dtype=float
)

UPPER_BOUNDS = np.array(
    [250, 20, 12, 6, 1.0] + [1.0] * N_FEATURES,
    dtype=float
)

def decode_particle(particle):

    p = np.clip(
        particle,
        LOWER_BOUNDS,
        UPPER_BOUNDS
    )

    mask = (p[5:] >= 0.5).astype(int)

    if mask.sum() == 0:
        mask[np.argmax(p[5:])] = 1

    selected = [
        feature_columns[i]
        for i in range(N_FEATURES)
        if mask[i] == 1
    ]

    parameters = {
        "n_estimators": int(round(p[0])),
        "max_depth": int(round(p[1])),
        "min_samples_split": int(round(p[2])),
        "min_samples_leaf": int(round(p[3])),
        "max_features": float(p[4]),
        "random_state": SEED,
        "n_jobs": 1
    }

    return parameters, selected, mask.tolist()

# ================================================================
# 5. FITNESS FUNCTION
# ================================================================

fitness_cache = {}

cv = KFold(
    n_splits=3,
    shuffle=True,
    random_state=SEED
)

def calculate_fitness(particle):

    params, selected, mask = decode_particle(
        particle
    )

    cache_key = (
        params["n_estimators"],
        params["max_depth"],
        params["min_samples_split"],
        params["min_samples_leaf"],
        round(params["max_features"], 4),
        tuple(mask)
    )

    if cache_key in fitness_cache:
        return fitness_cache[cache_key]

    pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "rf",
            RandomForestRegressor(**params)
        )
    ])

    try:
        scores = cross_val_score(
            pipeline,
            X_train_full[selected],
            y_train_full,
            cv=cv,
            scoring="neg_root_mean_squared_error",
            n_jobs=-1,
            error_score="raise"
        )

        rmse = float(-np.mean(scores))

    except Exception:
        rmse = 1e12

    fitness_cache[cache_key] = rmse

    return rmse

# ================================================================
# 6. INITIALIZE PARTICLE SWARM
# ================================================================

positions = rng.uniform(
    LOWER_BOUNDS,
    UPPER_BOUNDS,
    size=(NUM_PARTICLES, DIMENSIONS)
)

velocities = np.zeros_like(positions)

personal_best_positions = positions.copy()

personal_best_scores = np.array([
    calculate_fitness(p)
    for p in positions
])

best_index = np.argmin(
    personal_best_scores
)

global_best_position = (
    personal_best_positions[best_index].copy()
)

global_best_score = float(
    personal_best_scores[best_index]
)

convergence_history = [
    global_best_score
]

mean_fitness_history = [
    float(np.mean(personal_best_scores))
]

# ================================================================
# 7. PSO OPTIMIZATION LOOP
# ================================================================

print("\nStarting PSO optimization...\n")

velocity_limit = (
    UPPER_BOUNDS - LOWER_BOUNDS
) * 0.20

for iteration in range(
    1,
    MAX_ITERATIONS + 1
):

    iteration_scores = []

    for i in range(NUM_PARTICLES):

        r1 = rng.random(DIMENSIONS)
        r2 = rng.random(DIMENSIONS)

        # Standard PSO velocity update.
        velocities[i] = (
            INERTIA * velocities[i]
            + COGNITIVE * r1 * (
                personal_best_positions[i]
                - positions[i]
            )
            + SOCIAL * r2 * (
                global_best_position
                - positions[i]
            )
        )

        velocities[i] = np.clip(
            velocities[i],
            -velocity_limit,
            velocity_limit
        )

        # Position update.
        positions[i] = (
            positions[i] + velocities[i]
        )

        positions[i] = np.clip(
            positions[i],
            LOWER_BOUNDS,
            UPPER_BOUNDS
        )

        fitness = calculate_fitness(
            positions[i]
        )

        iteration_scores.append(fitness)

        # Update personal best.
        if fitness < personal_best_scores[i]:

            personal_best_scores[i] = fitness

            personal_best_positions[i] = (
                positions[i].copy()
            )

        # Update global best.
        if fitness < global_best_score:

            global_best_score = fitness

            global_best_position = (
                positions[i].copy()
            )

    convergence_history.append(
        global_best_score
    )

    mean_fitness_history.append(
        float(np.mean(iteration_scores))
    )

    print(
        f"Iteration {iteration:02d}/"
        f"{MAX_ITERATIONS} | "
        f"Best CV RMSE: "
        f"{global_best_score:.4f}"
    )

print("\nPSO optimization completed!")

# ================================================================
# 8. DECODE BEST SOLUTION
# ================================================================

best_params, selected_features, feature_mask = (
    decode_particle(global_best_position)
)

print("\nBest PSO hyperparameters:")
print(best_params)

print("\nSelected features:")
print(selected_features)

print("\nBest cross-validation RMSE:")
print(global_best_score)

# ================================================================
# 9. TRAIN PSO-OPTIMIZED RANDOM FOREST
# ================================================================

# Fit on inner training data for validation selection.
# Refit on full training data after selecting the model.

def make_rf_pipeline():

    params = best_params.copy()
    params["n_jobs"] = -1

    return Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "rf",
            RandomForestRegressor(**params)
        )
    ])

rf_validation_model = make_rf_pipeline()

rf_validation_model.fit(
    X_train[selected_features],
    y_train
)

rf_val_predictions = np.maximum(
    rf_validation_model.predict(
        X_val[selected_features]
    ),
    0
)

# ================================================================
# 10. TRAIN PSO-GUIDED NEURAL NETWORK
# ================================================================

# PSO selects the input features.
# Adam trains the neural network weights.

def make_nn_model(input_size):

    model = Sequential([
        Input(shape=(input_size,)),
        Dense(64, activation="relu"),
        Dropout(0.20),
        Dense(32, activation="relu"),
        Dropout(0.10),
        Dense(16, activation="relu"),
        Dense(1, activation="linear")
    ])

    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=0.001
        ),
        loss="mse",
        metrics=["mae"]
    )

    return model

imputer_val = SimpleImputer(
    strategy="median"
)

X_train_imp = imputer_val.fit_transform(
    X_train[selected_features]
)

X_val_imp = imputer_val.transform(
    X_val[selected_features]
)

scaler_val = StandardScaler()

X_train_scaled = scaler_val.fit_transform(
    X_train_imp
)

X_val_scaled = scaler_val.transform(
    X_val_imp
)

nn_validation_model = make_nn_model(
    len(selected_features)
)

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=20,
    restore_best_weights=True
)

history = nn_validation_model.fit(
    X_train_scaled,
    np.log1p(y_train.to_numpy()),
    validation_data=(
        X_val_scaled,
        np.log1p(y_val.to_numpy())
    ),
    epochs=200,
    batch_size=16,
    callbacks=[early_stopping],
    verbose=0,
    shuffle=True
)

nn_val_log = nn_validation_model.predict(
    X_val_scaled,
    verbose=0
).flatten()

nn_val_predictions = np.maximum(
    np.expm1(
        np.clip(nn_val_log, -20, 20)
    ),
    0
)

# ================================================================
# 11. MODEL EVALUATION FUNCTION
# ================================================================

def evaluate(actual, predicted):

    return {
        "MAE": float(
            mean_absolute_error(
                actual, predicted
            )
        ),
        "MSE": float(
            mean_squared_error(
                actual, predicted
            )
        ),
        "RMSE": float(
            np.sqrt(
                mean_squared_error(
                    actual, predicted
                )
            )
        ),
        "R2": float(
            r2_score(
                actual, predicted
            )
        )
    }

rf_val_metrics = evaluate(
    y_val.to_numpy(),
    rf_val_predictions
)

nn_val_metrics = evaluate(
    y_val.to_numpy(),
    nn_val_predictions
)

print("\nValidation performance:")
print("PSO Random Forest:", rf_val_metrics)
print("PSO-Guided NN:", nn_val_metrics)

if rf_val_metrics["RMSE"] <= nn_val_metrics["RMSE"]:

    best_model_name = "PSO-Random Forest"

else:

    best_model_name = "PSO-Guided Neural Network"

print("\nValidation-selected model:", best_model_name)

# ================================================================
# 12. REFIT RANDOM FOREST ON FULL TRAINING DATA
# ================================================================

pso_rf_model = make_rf_pipeline()

pso_rf_model.fit(
    X_train_full[selected_features],
    y_train_full
)

rf_predictions = np.maximum(
    pso_rf_model.predict(
        X_test[selected_features]
    ),
    0
)

# ================================================================
# 13. REFIT NEURAL NETWORK ON FULL TRAINING DATA
# ================================================================

imputer = SimpleImputer(
    strategy="median"
)

X_full_imp = imputer.fit_transform(
    X_train_full[selected_features]
)

X_test_imp = imputer.transform(
    X_test[selected_features]
)

scaler = StandardScaler()

X_full_scaled = scaler.fit_transform(
    X_full_imp
)

X_test_scaled = scaler.transform(
    X_test_imp
)

# Use the number of epochs selected by validation
# training, without reusing the test set.

best_epochs = (
    int(np.argmin(history.history["val_loss"])) + 1
)

tf.keras.backend.clear_session()
tf.random.set_seed(SEED)

pso_nn_model = make_nn_model(
    len(selected_features)
)

pso_nn_model.fit(
    X_full_scaled,
    np.log1p(y_train_full.to_numpy()),
    epochs=best_epochs,
    batch_size=16,
    verbose=0,
    shuffle=True
)

nn_log_predictions = pso_nn_model.predict(
    X_test_scaled,
    verbose=0
).flatten()

nn_predictions = np.maximum(
    np.expm1(
        np.clip(nn_log_predictions, -20, 20)
    ),
    0
)

# ================================================================
# 14. FINAL TEST EVALUATION
# ================================================================

actual = y_test.to_numpy()

rf_metrics = evaluate(
    actual,
    rf_predictions
)

nn_metrics = evaluate(
    actual,
    nn_predictions
)

comparison_df = pd.DataFrame([
    {
        "Model": "PSO-Random Forest",
        **rf_metrics
    },
    {
        "Model": "PSO-Guided Neural Network",
        **nn_metrics
    }
])

comparison_df.to_csv(
    out("model_comparison.csv"),
    index=False
)

if best_model_name == "PSO-Random Forest":

    best_predictions = rf_predictions
    best_metrics = rf_metrics

else:

    best_predictions = nn_predictions
    best_metrics = nn_metrics

print("\nFINAL TEST PERFORMANCE")
print(comparison_df.to_string(index=False))

print("\nSelected Model:", best_model_name)

# ================================================================
# 15. SAVE MODEL ARTIFACTS
# ================================================================

joblib.dump(
    {
        "pipeline": pso_rf_model,
        "selected_features": selected_features,
        "target": TARGET
    },
    out("vegetable_model.pkl")
)

joblib.dump(
    scaler,
    out("vegetable_scaler.pkl")
)

joblib.dump(
    imputer,
    out("vegetable_imputer.pkl")
)

pso_nn_model.save(
    str(out("vegetable_model.h5"))
)

print("\nPSO model files saved successfully!")

# ================================================================
# 16. SAVE YAML CONFIGURATION
# ================================================================

config = {
    "project_name": (
        "AI-Based Vegetable Crop Production Analysis"
    ),
    "optimization": "Particle Swarm Optimization",
    "algorithm": "PSO",
    "dataset": DATASET_PATH.name,
    "target": TARGET,
    "target_unit": "tonnes",
    "selected_features": selected_features,
    "feature_mask": feature_mask,
    "random_state": SEED,
    "data_split": {
        "outer_test_size": 0.20,
        "inner_validation_size": 0.20
    },
    "pso_parameters": {
        "particles": NUM_PARTICLES,
        "iterations": MAX_ITERATIONS,
        "inertia": INERTIA,
        "cognitive_coefficient": COGNITIVE,
        "social_coefficient": SOCIAL
    },
    "best_rf_parameters": best_params,
    "best_cv_rmse": float(global_best_score),
    "nn_epochs": best_epochs,
    "validation_selected_model": best_model_name
}

with open(
    out("config.yaml"),
    "w",
    encoding="utf-8"
) as f:

    yaml.safe_dump(
        config,
        f,
        sort_keys=False
    )

# ================================================================
# 17. SAVE JSON METADATA
# ================================================================

metadata = {
    "project": (
        "AI-Based Vegetable Crop Production Analysis"
    ),
    "optimization": "PSO",
    "created_at": datetime.now().isoformat(),
    "dataset": DATASET_PATH.name,
    "target": TARGET,
    "target_unit": "tonnes",
    "selected_features": selected_features,
    "best_parameters": best_params,
    "best_cv_rmse": float(global_best_score),
    "validation_metrics": {
        "PSO-Random Forest": rf_val_metrics,
        "PSO-Guided Neural Network": nn_val_metrics
    },
    "test_metrics": {
        "PSO-Random Forest": rf_metrics,
        "PSO-Guided Neural Network": nn_metrics
    },
    "selected_model": best_model_name,
    "convergence_history": convergence_history,
    "mean_fitness_history": mean_fitness_history,
    "training_samples": int(len(X_train_full)),
    "testing_samples": int(len(X_test)),
    "neural_network_epochs": best_epochs,
    "limitations": [
        "Single-year crop production dataset",
        "Not a future-year forecasting model",
        "PSO optimizes RF parameters and feature selection",
        "Neural network uses PSO-selected features",
        "Regional production is not equivalent to yield per hectare"
    ]
}

with open(
    out("model_metadata.json"),
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        metadata,
        f,
        indent=4
    )

# ================================================================
# 18. ACCURACY GRAPH - R2 SCORE
# ================================================================

names = [
    "PSO-Random Forest",
    "PSO-Guided NN"
]

scores = [
    rf_metrics["R2"],
    nn_metrics["R2"]
]

fig, ax = plt.subplots(figsize=(10, 6))

bars = ax.bar(
    names,
    scores,
    color=["#3498db", "#2ecc71"],
    width=0.55
)

ax.axhline(
    0,
    color="black",
    linewidth=1
)

for bar, score in zip(bars, scores):

    ax.annotate(
        f"{score:.4f}",
        (
            bar.get_x() + bar.get_width() / 2,
            bar.get_height()
        ),
        xytext=(0, 5 if score >= 0 else -15),
        textcoords="offset points",
        ha="center",
        fontsize=11,
        fontweight="bold"
    )

ax.set_title(
    "PSO Model Performance (R2 Score)",
    fontweight="bold"
)

ax.set_xlabel("Models")
ax.set_ylabel("R2 Score")
ax.grid(axis="y", alpha=0.3)

save_plot("accuracy_graph.png")

# ================================================================
# 19. CORRELATION HEATMAP
# ================================================================

fig, ax = plt.subplots(figsize=(13, 10))

sns.heatmap(
    df[crop_columns].corr(),
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    center=0,
    linewidths=0.5,
    square=True,
    ax=ax
)

ax.set_title(
    "PSO Study - Vegetable Production Correlation",
    fontsize=15,
    fontweight="bold"
)

plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)

save_plot("heatmap.png")

# ================================================================
# 20. MODEL COMPARISON GRAPH
# ================================================================

metrics = ["MAE", "RMSE"]

x = np.arange(len(metrics))
width = 0.35

rf_values = [
    rf_metrics[m] for m in metrics
]

nn_values = [
    nn_metrics[m] for m in metrics
]

fig, ax = plt.subplots(figsize=(11, 7))

bars1 = ax.bar(
    x - width / 2,
    rf_values,
    width,
    label="PSO-Random Forest",
    color="#3498db"
)

bars2 = ax.bar(
    x + width / 2,
    nn_values,
    width,
    label="PSO-Guided NN",
    color="#2ecc71"
)

ax.set_xticks(x)
ax.set_xticklabels(metrics)

ax.set_title(
    "PSO Model Comparison",
    fontweight="bold"
)

ax.set_ylabel("Prediction Error (Tonnes)")
ax.legend()
ax.grid(axis="y", alpha=0.3)

for bars in [bars1, bars2]:

    for bar in bars:

        ax.annotate(
            f"{bar.get_height():.2f}",
            (
                bar.get_x() + bar.get_width() / 2,
                bar.get_height()
            ),
            xytext=(0, 4),
            textcoords="offset points",
            ha="center",
            fontsize=9
        )

save_plot("comparison_graph.png")

# ================================================================
# 21. GENERATE RESULT CSV
# ================================================================

test_indices = X_test.index

if LOCATION_COLUMN in df.columns:

    taluks = df.loc[
        test_indices,
        LOCATION_COLUMN
    ].values

else:

    taluks = test_indices.to_numpy()

result_df = pd.DataFrame({
    "Taluk": taluks,
    "Actual_Tomato_Production": actual,
    "PSO_RF_Prediction": rf_predictions,
    "PSO_NN_Prediction": nn_predictions,
    "PSO_RF_Absolute_Error": np.abs(
        actual - rf_predictions
    ),
    "PSO_NN_Absolute_Error": np.abs(
        actual - nn_predictions
    ),
    "Selected_Model": best_model_name,
    "Selected_Model_Prediction": best_predictions,
    "Selected_Model_Absolute_Error": np.abs(
        actual - best_predictions
    )
})

result_df.to_csv(
    out("result.csv"),
    index=False
)

print("\nPSO Result CSV:")
print(result_df.head())

# ================================================================
# 22. RESULT GRAPH
# ================================================================

sorted_results = result_df.sort_values(
    "Actual_Tomato_Production"
).reset_index(drop=True)

fig, ax = plt.subplots(figsize=(14, 7))

ax.plot(
    sorted_results.index,
    sorted_results["Actual_Tomato_Production"],
    label="Actual Production",
    color="black",
    linewidth=2.5
)

ax.plot(
    sorted_results.index,
    sorted_results["PSO_RF_Prediction"],
    label="PSO-Random Forest",
    color="#3498db",
    linestyle="--",
    linewidth=2
)

ax.plot(
    sorted_results.index,
    sorted_results["PSO_NN_Prediction"],
    label="PSO-Guided NN",
    color="#2ecc71",
    linestyle=":",
    linewidth=2
)

ax.set_title(
    "PSO Actual vs Estimated Tomato Production",
    fontweight="bold"
)

ax.set_xlabel(
    "Test Samples Sorted by Actual Production"
)

ax.set_ylabel("Tomato Production (Tonnes)")
ax.legend()
ax.grid(alpha=0.3)

save_plot("result_graph.png")

# ================================================================
# 23. GENERATE PREDICTION CSV
# ================================================================

prediction_df = X_test.copy()

prediction_df.insert(
    0,
    "Taluk",
    taluks
)

prediction_df["Actual_Tomato_Production"] = (
    actual
)

prediction_df["Predicted_Tomato_Production"] = (
    best_predictions
)

prediction_df["Prediction_Error"] = (
    actual - best_predictions
)

prediction_df["Absolute_Error"] = np.abs(
    prediction_df["Prediction_Error"]
)

prediction_df["Model_Used"] = best_model_name

prediction_df.to_csv(
    out("prediction.csv"),
    index=False
)

print("\nPSO Prediction CSV:")
print(prediction_df.head())

# ================================================================
# 24. PREDICTION GRAPH
# ================================================================

fig, ax = plt.subplots(figsize=(11, 8))

ax.scatter(
    actual,
    best_predictions,
    color="#2980b9",
    alpha=0.75,
    s=65,
    edgecolors="black",
    linewidths=0.4
)

minimum = min(
    actual.min(),
    best_predictions.min()
)

maximum = max(
    actual.max(),
    best_predictions.max()
)

ax.plot(
    [minimum, maximum],
    [minimum, maximum],
    "r--",
    linewidth=2,
    label="Perfect Prediction"
)

ax.set_title(
    f"PSO Prediction Performance - {best_model_name}",
    fontweight="bold"
)

ax.set_xlabel("Actual Tomato Production (Tonnes)")
ax.set_ylabel("Predicted Tomato Production (Tonnes)")

ax.text(
    0.05,
    0.95,
    f"R2 = {best_metrics['R2']:.4f}\n"
    f"RMSE = {best_metrics['RMSE']:.2f}\n"
    f"MAE = {best_metrics['MAE']:.2f}",
    transform=ax.transAxes,
    verticalalignment="top",
    bbox={
        "boxstyle": "round",
        "facecolor": "white",
        "alpha": 0.85
    }
)

ax.legend()
ax.grid(alpha=0.3)

save_plot("prediction_graph.png")

# ================================================================
# 25. FEATURE IMPORTANCE GRAPH
# ================================================================

importance = pso_rf_model.named_steps[
    "rf"
].feature_importances_

importance_df = pd.DataFrame({
    "Feature": selected_features,
    "Importance": importance
}).sort_values("Importance")

fig, ax = plt.subplots(figsize=(11, 7))

ax.barh(
    importance_df["Feature"],
    importance_df["Importance"],
    color="#9b59b6"
)

ax.set_title(
    "PSO-Selected Feature Importance",
    fontweight="bold"
)

ax.set_xlabel("Random Forest Feature Importance")

save_plot("feature_importance.png")

# ================================================================
# 26. PSO CONVERGENCE GRAPH
# ================================================================

fig, ax = plt.subplots(figsize=(11, 7))

ax.plot(
    range(len(convergence_history)),
    convergence_history,
    marker="o",
    color="#e67e22",
    linewidth=2,
    label="Global Best CV RMSE"
)

ax.plot(
    range(len(mean_fitness_history)),
    mean_fitness_history,
    linestyle="--",
    color="#3498db",
    alpha=0.75,
    label="Mean Swarm CV RMSE"
)

ax.set_title(
    "PSO Optimization Convergence",
    fontweight="bold"
)

ax.set_xlabel("Iteration")
ax.set_ylabel("Cross-Validation RMSE (Tonnes)")

ax.legend()
ax.grid(alpha=0.3)

save_plot("convergence_graph.png")

# ================================================================
# 27. VERIFY ALL GENERATED FILES
# ================================================================

output_files = [
    "vegetable_model.h5",
    "vegetable_model.pkl",
    "vegetable_scaler.pkl",
    "vegetable_imputer.pkl",
    "config.yaml",
    "model_metadata.json",
    "accuracy_graph.png",
    "heatmap.png",
    "comparison_graph.png",
    "result.csv",
    "result_graph.png",
    "prediction.csv",
    "prediction_graph.png",
    "feature_importance.png",
    "convergence_graph.png",
    "model_comparison.csv"
]

print("\n" + "=" * 70)
print("PSO GENERATED OUTPUT FILES")
print("=" * 70)

for filename in output_files:

    path = out(filename)

    if path.exists():

        size_kb = path.stat().st_size / 1024

        print(
            f"[SUCCESS] {path.name} "
            f"({size_kb:.2f} KB)"
        )

    else:

        print(f"[FAILED] {path.name}")

print("\n" + "=" * 70)
print("PSO OPTIMIZATION COMPLETED")
print("=" * 70)

print("\nSelected Model:", best_model_name)
print("Best CV RMSE:", round(global_best_score, 4))
print("Test R2:", round(best_metrics["R2"], 4))
print("Test RMSE:", round(best_metrics["RMSE"], 4))
print("Test MAE:", round(best_metrics["MAE"], 4))

print("\nAll PSO files saved in:")
print(PROJECT_DIR)
```
